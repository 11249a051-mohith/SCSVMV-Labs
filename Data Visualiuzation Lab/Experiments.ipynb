```r


boxplot(
  mpg ~ cyl,
  data = mtcars,
  main = "Box Plot (MPG by Cyl)",
  xlab = "Number of Cylinders",
  ylab = "Miles per Gallon (MPG)",
  col = c("orange", "green", "blue")
)
```

```r
# Calculate correlation between the first 6 columns
res <- cor(mtcars[, 1:6])

# Draw the correlation heatmap
heatmap(res,
        main = "Correlation Heatmap",
        col = cm.colors(256))
```

```r
# Calculate the density of MPG values
d <- density(mtcars$mpg)

# Draw the density plot
plot(d, main = "Density Plot of MPG")

# Fill the area under the curve with purple
polygon(d, col = "purple", border = "black")
```

```r
# Aim: To visualize correlations between variables

# Calculate correlation for the first 6 columns
res <- cor(mtcars[, 1:6])

# Display the correlation heatmap
heatmap(res,
        main = "Correlation Heatmap",
        col = cm.colors(256))

# Result: Correlation heatmap displayed successfully
```

```r
# Aim: To visualize the distribution of MPG values

d <- density(mtcars$mpg)

plot(d, main = "Density Plot of MPG")

polygon(d, col = "purple", border = "black")

# Result: Density plot displayed successfully
```

```r
# Aim: To visualize the distribution of numerical data

data <- rnorm(100)

hist(data,
     col = "skyblue",
     border = "white",
     main = "Histogram Example",
     xlab = "Value")

# Result: Histogram displayed successfully
```

```r
# Aim: To compare MPG for cars with different numbers of cylinders

boxplot(mpg ~ cyl,
        data = mtcars,
        main = "Boxplot: MPG vs Cylinders",
        col = c("gold", "coral", "steelblue"),
        xlab = "Cylinders",
        ylab = "MPG")

# Result: Boxplot displayed successfully
```

```r
# Aim: To visualize correlations between all variables in mtcars

# Calculate the correlation matrix
res <- cor(mtcars)

# Display the heatmap
image(res,
      main = "Correlation Matrix Heatmap",
      axes = FALSE)

# Add X-axis labels
axis(1,
     at = seq(0, 1, length.out = ncol(mtcars)),
     labels = colnames(mtcars),
     las = 2)

# Add Y-axis labels
axis(2,
     at = seq(0, 1, length.out = ncol(mtcars)),
     labels = colnames(mtcars),
     las = 2)

# Result: Correlation matrix heatmap displayed successfully
```

```r
# Aim: To visualize the distribution of MPG values

# Calculate kernel density
d <- density(mtcars$mpg)

# Plot the density curve
plot(d, main = "Kernel Density of MPG")

# Fill the area under the curve
polygon(d, col = "red", border = "blue")

# Result: Kernel density plot displayed successfully
```
