```r
x <- c(1, 2, 3, 4, 5)
y <- c(10, 15, 12, 18, 20)
plot(x, y, type="o",
     main="Line Graph",
     xlab="X Values",
     ylab="Y Values")
```

```r
subjects <- c("Python", "Java", "R", "Tableau", "SQL")
marks <- c(85, 78, 82, 90, 88)
barplot(marks,
        names.arg=subjects,
        main="Marks by Subject",
        xlab="Subject",
        ylab="Marks")
```

```r
values <- c(35, 25, 20, 20)
labels <- c("Python", "Java", "R", "Tableau")

pie(values,
    labels=labels,
    main="Technology Usage")
```

```r
hours <- c(1, 2, 3, 4, 5, 6)
marks <- c(45, 50, 58, 65, 72, 80)
plot(hours, marks,
     main="Study Hours vs Marks",
     xlab="Study Hours",
     ylab="Marks",
     pch=19)
```

```r
marks <- c(45, 50, 52, 55, 58, 60, 62, 65,
           68, 70, 72, 75, 78, 80, 82, 85,
           88, 90)
hist(marks,
     breaks=5,
     main="Distribution of Marks",
     xlab="Marks",
     ylab="Frequency")
```
